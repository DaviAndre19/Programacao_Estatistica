```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def exponencial_inversa(lam):
  u = np.random.uniform()
  assert lam>0, "Lambda n eh maior que 0"
  x = (1/(-lam))*(np.log(1-u))

  return x
```

```python
x = exponencial_inversa(2)
print(x)
```

```python
amostra_exponencial=[]
for _ in range(1000):
  k = exponencial_inversa(2)
  amostra_exponencial.append(float(k))
print(amostra_exponencial)
```

```python
plt.hist(amostra_exponencial)
```

```python
def geometrica(p):
  u = np.random.uniform()
  x=np.floor((np.log(1-u)/np.log(1-p)))+1
  return x
```

```python
def exponencial_geo(n,lam):
  t=geometrica(lam/n)
  x1= t/n
  return x1
```

```python
def exponencial_poisson(u):
  tempo = 0
  contagem = 0
  while tempo<1:
    tempo+=exponencial_inversa(u)
    contagem+=1
  return contagem - 1
```

```python
amostra_poisson =[]
for _ in range(10**4):
  poisson = exponencial_poisson(4)
  amostra_poisson.append(poisson)
print(amostra_poisson)
```

```python
amostra2_poisson=[]
for _ in range(10**4):
  poisson2 = np.random.poisson(4)
  amostra2_poisson.append(poisson2)
print(amostra2_poisson)
```

```python
plt.hist(amostra_poisson, label = 'Amostra criada', color = "red")
plt.hist(amostra2_poisson, label = "Poisson do Python", color = "blue")
plt.legend()
plt.show()
```
